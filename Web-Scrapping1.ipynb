```python
import os
import time
import pandas as pd
import requests
from bs4 import BeautifulSoup

BASE_URL = "https://ambitionbox.com"

session = requests.Session()
session.headers.update({
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36',
    'Accept-Language': 'en-US,en;q=0.9',
    'Referer': BASE_URL + '/'
})

csv_filename = 'ambitionbox_full_dataset.csv'

# Automatically resume from the next page if the CSV already exists
if os.path.isfile(csv_filename):
    existing_df = pd.read_csv(csv_filename)
    start_page = int(existing_df['Page'].max()) + 1 if not existing_df.empty else 1
    print(f"Resuming automatically from Page {start_page} to 500...")
else:
    start_page = 1

# Loop cleanly from start_page up to page 500
for page in range(start_page, 501):
    print(f"\n========== Scraping Page {page} of 500 ==========")
    url = BASE_URL + "/list-of-companies?campaign=desktop_nav&page=" + str(page)
    
    try:
        res = session.get(url, timeout=15)
    except Exception as e:
        print(f"Network error on page {page}: {e}. Re-run cell to resume safely.")
        break

    if res.status_code != 200:
        print(f"Stopped at page {page} (Status Code: {res.status_code}). Wait a few minutes and re-run to resume.")
        break

    soup = BeautifulSoup(res.text, 'html.parser')

    # 1. Extract main page columns
    name = [i.text.strip() for i in soup.find_all('h2')][:20]
    if len(name) == 0:
        print(f"No more companies found on page {page}. Stopping.")
        break

    ratings = [i.text.strip() for i in soup.find_all('span', class_='text-primary-text font-pn-700 text-sm !text-[14px] !leading-[14px]')][:20]
    reviews = [i.text.strip() for i in soup.find_all('span', class_='pl-1 text-[14px] font-pn-400 leading-5 text-[#4E597B] max-md:text-[12px]')][:20]

    company_types = []
    cities = []
    for i in soup.find_all('span', class_='overflow-hidden text-left text-[14px] font-pn-400 leading-5 tracking-[0.25px] text-[#394468] max-md:text-[12px] max-md:leading-4')[:20]:
        full_text = i.text.strip()
        company_types.append(full_text.split('|')[0].strip())
        cities.append(full_text.split('|')[1].strip() if '|' in full_text else None)

    # 2. Highly Rated For & Critically Rated For
    highly_rated = []
    critically_rated = []
    for h2 in soup.find_all('h2')[:20]:
        card = h2.parent
        while card and 'Reviews' not in card.get_text():
            card = card.parent

        high_val, crit_val = None, None
        if card:
            spans = card.find_all('span', class_='text-left text-[14px] font-pn-400 leading-5 tracking-[0.25px] text-[#1E223C] max-md:text-[12px] max-md:leading-4')
            for s in spans:
                prev_text = s.find_previous('span').text if s.find_previous('span') else ''
                if 'Highly Rated For' in prev_text:
                    high_val = s.text.strip()
                elif 'Critically Rated For' in prev_text:
                    crit_val = s.text.strip()
        highly_rated.append(high_val)
        critically_rated.append(crit_val)

    # 3. Extract review links for this page
    links = []
    for a in soup.find_all('a', href=True):
        href = a['href'].split('?')[0]
        if '/reviews/' in href and href.endswith('-reviews'):
            full_url = href if href.startswith('http') else BASE_URL + href
            if full_url not in links:
                links.append(full_url)

    # 4. Visit each link to get Women & Men reviews and ratings
    women_reviews, women_ratings = [], []
    men_reviews, men_ratings = [], []

    for idx in range(len(name)):
        link = links[idx] if idx < len(links) else None
        w_count, w_rating, m_count, m_rating = None, None, None, None

        if link:
            try:
                page_res = session.get(link, timeout=15)
                if page_res.status_code == 200:
                    page_soup = BeautifulSoup(page_res.text, 'html.parser')

                    female_card = page_soup.find('div', {'data-testid': 'genderInsightsCard_female'})
                    w_count = female_card.find('span', {'data-testid': 'count'}).text.strip('() ') if female_card else None
                    w_rating = female_card.find('span', {'data-testid': 'rating'}).text.strip() if female_card else None

                    male_card = page_soup.find('div', {'data-testid': 'genderInsightsCard_male'})
                    m_count = male_card.find('span', {'data-testid': 'count'}).text.strip('() ') if male_card else None
                    m_rating = male_card.find('span', {'data-testid': 'rating'}).text.strip() if male_card else None
            except Exception:
                pass

            time.sleep(1.5)

        women_reviews.append(w_count)
        women_ratings.append(w_rating)
        men_reviews.append(m_count)
        men_ratings.append(m_rating)
        print(f"Page {page} | {idx+1}. {name[idx]} -> Women: {w_count} ({w_rating}) | Men: {m_count} ({m_rating})")

    # 5. Save this page immediately to CSV
    page_df = pd.DataFrame({
        'Page': [page] * len(name),
        'Company Name': name,
        'Overall Rating': ratings + [None] * (len(name) - len(ratings)),
        'Total Reviews': reviews + [None] * (len(name) - len(reviews)),
        'Industry': company_types + [None] * (len(name) - len(company_types)),
        'City': cities + [None] * (len(name) - len(cities)),
        'Women Reviews': women_reviews,
        'Women Rating': women_ratings,
        'Men Reviews': men_reviews,
        'Men Rating': men_ratings,
        'Highly Rated For': highly_rated,
        'Critically Rated For': critically_rated
    })

    file_exists = os.path.isfile(csv_filename)
    page_df.to_csv(csv_filename, mode='a', header=not file_exists, index=False)
    print(f"--> Page {page} saved to {csv_filename}")

    time.sleep(2)

# Load your complete saved CSV into df once finished
if os.path.isfile(csv_filename):
    df = pd.read_csv(csv_filename)
    print(f"\nTotal companies saved in CSV: {len(df)}")
    display(df.head())
```
